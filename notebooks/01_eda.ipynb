```python
import pandas as pd

DATA_PATH = "../data/raw/diabetic_data.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
df["readmitted_30"] = (df["readmitted"] == "<30").astype(int)

print("Target created successfully")
```

```python
print("Original Target:")
print(df["readmitted"].value_counts())

print("\n30-Day Binary Target:")
print(df["readmitted_30"].value_counts())

print("\n30-Day Binary Target Percentage:")
print(df["readmitted_30"].value_counts(normalize=True) * 100)
```

```python
print("Original Target:")
print(df["readmitted"].value_counts())

print("\n30-Day Binary Target:")
print(df["readmitted_30"].value_counts())

print("\n30-Day Binary Target Percentage:")
print(df["readmitted_30"].value_counts(normalize=True) * 100)
```

```python
print("Number of columns:", len(df.columns))

print("\nColumns:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

```python
df.info()
```

```python
missing_q = (df == "?").sum().sort_values(ascending=False)

print("Top columns containing '?' values:")
print(missing_q.head(20))
```

```python
missing_percentage = (df == "?").mean().sort_values(ascending=False) * 100

print("Top columns by '?' percentage:")
print(missing_percentage.head(20))
```

```python
print("Duplicate rows:", df.duplicated().sum())

print("Unique encounters:", df["encounter_id"].nunique())

print("Total rows:", len(df))

print("Unique patients:", df["patient_nbr"].nunique())
```

```python
df.describe(include="all").T
```

```python
print("===== TARGET DISTRIBUTION =====")

print("\nReadmission counts:")
print(df["readmitted"].value_counts())

print("\nReadmission percentages:")
print(df["readmitted"].value_counts(normalize=True).mul(100).round(2))
```

```python
print("===== MISSING VALUES =====")

missing_q = (df == "?").sum().sort_values(ascending=False)

missing_q = missing_q[missing_q > 0]

missing_summary = pd.DataFrame({
    "missing_count": missing_q,
    "missing_percentage": (missing_q / len(df) * 100).round(2)
})

missing_summary
```

```python
print("===== IDENTIFIERS =====")

print("Total encounters:", len(df))
print("Unique encounters:", df["encounter_id"].nunique())
print("Unique patients:", df["patient_nbr"].nunique())
print("Duplicate rows:", df.duplicated().sum())
```

```python
patient_counts = df["patient_nbr"].value_counts()

print("Patients with multiple encounters:",
      (patient_counts > 1).sum())

print("Maximum encounters for one patient:",
      patient_counts.max())
```
