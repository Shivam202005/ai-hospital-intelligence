# Member 2: EDA Review and Feature Analysis

This section extends the initial EDA with demographic distributions, numerical feature distributions, categorical feature review, feature interpretation, and feature relationships with the 30-day readmission target.

```python
demographic_features = ["race", "gender", "age", "weight"]

for col in demographic_features:
    print(f"\n===== {col.upper()} =====")
    print(df[col].value_counts(dropna=False))
```

```python
# Numerical feature distributions

numeric_features = [
    "time_in_hospital",
    "num_lab_procedures",
    "num_procedures",
    "num_medications",
    "number_outpatient",
    "number_emergency",
    "number_inpatient",
    "number_diagnoses"
]

print("===== NUMERICAL FEATURE SUMMARY =====")
print(df[numeric_features].describe().T)
```

```python
# Categorical feature review

categorical_features = [
    "race",
    "gender",
    "age",
    "weight",
    "payer_code",
    "medical_specialty",
    "max_glu_serum",
    "A1Cresult",
    "change",
    "diabetesMed"
]

print("===== CATEGORICAL FEATURE REVIEW =====")

for col in categorical_features:
    print(f"\n===== {col.upper()} =====")
    print(df[col].value_counts(dropna=False).head(15))
```

## Feature Interpretation

The following features were reviewed based on their meaning and potential relevance to 30-day hospital readmission prediction.

| Feature | Meaning / Interpretation | Potential Relevance |
|---|---|---|
| `age` | Patient age group | Demographic risk factor |
| `gender` | Patient gender | Demographic characteristic |
| `race` | Patient race | Demographic characteristic |
| `time_in_hospital` | Length of the current hospital stay | May indicate hospitalization complexity |
| `num_lab_procedures` | Number of laboratory procedures during the encounter | Indicates clinical activity |
| `num_procedures` | Number of procedures during the encounter | Indicates treatment intensity |
| `num_medications` | Number of medications during the encounter | May indicate treatment complexity |
| `number_outpatient` | Number of previous outpatient visits | Indicates prior healthcare utilization |
| `number_emergency` | Number of previous emergency visits | Indicates prior acute healthcare utilization |
| `number_inpatient` | Number of previous inpatient visits | Indicates prior hospitalization history |
| `number_diagnoses` | Number of diagnoses recorded | May indicate clinical complexity |
| `max_glu_serum` | Maximum glucose test result category | May indicate diabetes control |
| `A1Cresult` | HbA1c test result category | May indicate diabetes control |
| `change` | Whether diabetes medications were changed | Indicates treatment changes |
| `diabetesMed` | Whether diabetes medication was prescribed | Indicates diabetes treatment status |

These features are considered potential candidates for further analysis. Final feature selection and preprocessing will be decided during the modeling stage.


```python
# Numerical features vs 30-day readmission

print("===== NUMERICAL FEATURES VS 30-DAY READMISSION =====")

target_summary = df.groupby("readmitted_30")[numeric_features].mean().T

print(target_summary)
```

```python
print("===== MEDIAN VALUES BY READMISSION STATUS =====")

median_summary = df.groupby("readmitted_30")[numeric_features].median().T

print(median_summary)
```

```python
# Demographic features vs 30-day readmission

print("===== AGE VS 30-DAY READMISSION =====")
age_readmission = pd.crosstab(
    df["age"],
    df["readmitted_30"],
    normalize="index"
).round(3)

print(age_readmission)

print("\n===== GENDER VS 30-DAY READMISSION =====")
gender_readmission = pd.crosstab(
    df["gender"],
    df["readmitted_30"],
    normalize="index"
).round(3)

print(gender_readmission)

print("\n===== RACE VS 30-DAY READMISSION =====")
race_readmission = pd.crosstab(
    df["race"],
    df["readmitted_30"],
    normalize="index"
).round(3)

print(race_readmission)
```

## EDA Findings and Feature Observations

### Numerical Features
Patients readmitted within 30 days generally had higher values for:
- Previous inpatient visits
- Previous emergency visits
- Number of medications
- Number of diagnoses
- Time spent in the hospital
- Number of laboratory procedures

The largest difference was observed in `number_inpatient`, suggesting that previous inpatient utilization may be an important predictor of 30-day readmission.

### Demographic Features
Age showed some variation in readmission rates across age groups, while gender showed very little difference between male and female patients.

Race also showed some variation, but the differences were relatively small.

### Potentially Useful Features
Based on this EDA, the following features appear potentially useful for predicting 30-day readmission:

- `number_inpatient`
- `number_emergency`
- `number_outpatient`
- `time_in_hospital`
- `num_medications`
- `number_diagnoses`
- `num_lab_procedures`
- `age`
- `A1Cresult`
- `max_glu_serum`
- `change`
- `diabetesMed`

These are candidate features based on exploratory analysis and should not be considered final feature selections yet.

### Important Considerations
Features should be reviewed based on when the prediction is intended to be made. In particular, variables such as `discharge_disposition_id` may contain information that is only known later in the hospital encounter.

Further preprocessing, encoding, missing-value handling, and final feature selection should be decided as a team before model development.
